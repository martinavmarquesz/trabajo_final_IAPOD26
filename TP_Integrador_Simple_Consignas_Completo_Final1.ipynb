## Datos del trabajo

Vamos a usar una lista de diccionarios.  
Cada diccionario representa a un estudiante.

```python
estudiantes = [
    {"nombre": "Ana", "edad": 22, "trabaja": "Sí", "horas_estudio": 5, "nota": 8, "usa_ia": "Sí"},
    {"nombre": "Juan", "edad": 28, "trabaja": "Sí", "horas_estudio": 2, "nota": 5, "usa_ia": "No"},
    {"nombre": "Sofía", "edad": 20, "trabaja": "No", "horas_estudio": 6, "nota": 9, "usa_ia": "Sí"},
    {"nombre": "Pedro", "edad": 24, "trabaja": "No", "horas_estudio": 3, "nota": 6, "usa_ia": "No"},
    {"nombre": "Lucía", "edad": 31, "trabaja": "Sí", "horas_estudio": 4, "nota": 7, "usa_ia": "Sí"},
    {"nombre": "Martín", "edad": 19, "trabaja": "No", "horas_estudio": 2, "nota": 4, "usa_ia": "No"},
    {"nombre": "Carla", "edad": 26, "trabaja": "Sí", "horas_estudio": 7, "nota": 9, "usa_ia": "Sí"},
    {"nombre": "Diego", "edad": 23, "trabaja": "No", "horas_estudio": 4, "nota": 7, "usa_ia": "Sí"},
]
```

## Actividad 1: convertir los datos en una tabla

1.   Elemento de la lista
2.   Elemento de la lista



Importá pandas y convertí la lista `estudiantes` en un DataFrame llamado `df`.

Luego mostrá la tabla.

```python
import pandas as pd
import matplotlib as plt
df = pd.DataFrame(estudiantes)

print ("Tabla de estudiantes")
print(df)
```

## Actividad 2: calcular indicadores generales

Calculá y mostrá:

1. cantidad total de estudiantes;
2. promedio de edad;
3. promedio de nota;
4. promedio de horas de estudio.

```python
cantidad_estudiantes = len(df)
promedio_edad = df["edad"].mean()
promedio_nota = df ["nota"].mean()
promedio_horas = df["horas_estudio"].mean()

print( "Indicadores")
print(f"Cantidad total de estudiantes: {cantidad_estudiantes}")
print(f"Promedio de edad: {promedio_edad:.1f} años")
print(f"Promedio de nota: {promedio_nota:.2f}")
print(f"Promedio de horas de estudio: {promedio_horas:.1f} horas")

```

## Actividad 3: aprobados y desaprobados

Consideramos aprobado a quien tiene nota igual o mayor a 6.

Creá una nueva columna llamada `estado`, con los valores:

- `Aprobado`
- `Desaprobado`

Luego contá cuántos estudiantes aprobaron y cuántos desaprobaron.

```python

df["estado"] = df["nota"].apply(lambda x: "Aprobado" if x >= 6 else "Desaprobado")

conteo_estado = df["estado"].value_counts()

print(" Aprobados y desaprobados")
print(conteo_estado)
print(f"\nAprobados: {conteo_estado.get('Aprobado', 0)}")
print(f"Desaprobados: {conteo_estado.get('Desaprobado', 0)}")


print("Tabla con estado ")
print(df[["nombre", "nota", "estado"]])
```

## Actividad 4: trabajo y rendimiento

Separá los estudiantes en dos grupos:

- quienes trabajan;
- quienes no trabajan.

Luego calculá el promedio de nota de cada grupo.

```python
trabajan = df[df["trabaja"] == "Sí"]
no_trabajan = df[df["trabaja"] == "No"]

promedio_nota_trabajan = trabajan["nota"].mean()
promedio_nota_no_trabajan = no_trabajan["nota"].mean()

print(" Trabajo y rendimiento ")
print(f"Estudiantes que trabajan: {len(trabajan)}")
print(f"  Promedio de nota: {promedio_nota_trabajan:.2f}")
print(f"\nEstudiantes que NO trabajan: {len(no_trabajan)}")
print(f"  Promedio de nota: {promedio_nota_no_trabajan:.2f}")

```

## Actividad 5: uso de IA

Contá:

- cuántos estudiantes usan IA;
- cuántos estudiantes no usan IA.

```python
conteo_ia = df["usa_ia"].value_counts()

print(" Uso de inteligencia artificial ")
print(f"Usan IA: {conteo_ia.get('Sí', 0)} estudiantes")
print(f"No usan IA: {conteo_ia.get('No', 0)} estudiantes")


```

## Actividad 6: nota más alta y nota más baja

Mostrá:

- el nombre del estudiante con la nota más alta;
- el nombre del estudiante con la nota más baja.

```python
fila_max = df.loc[df["nota"].idxmax()]
fila_min = df.loc[df["nota"].idxmin()]

print(" Nota más alta y más baja ")
print(f"Nota más alta: {fila_max['nombre']} con {fila_max['nota']}")
print(f"Nota más baja: {fila_min['nombre']} con {fila_min['nota']}")


```

## Actividad 7: gráficos

Realizá al menos dos gráficos:

1. gráfico de barras con las notas de cada estudiante;
2. gráfico de dispersión con horas de estudio y nota final.

Podés usar `matplotlib`.

```python

import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 5))

colores = ["green" if nota >= 6 else "red" for nota in df["nota"]]
ax.bar(df["nombre"], df["nota"], color=colores, edgecolor="black")

ax.axhline(y=6, color="orange", linestyle="--", linewidth=1.5, label="Mínimo aprobado (6)")
ax.set_title("Notas por estudiante", fontsize=14, fontweight="bold")
ax.set_xlabel("Estudiante")
ax.set_ylabel("Nota")
ax.set_ylim(0, 10)
ax.legend()
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(df["horas_estudio"], df["nota"], color="steelblue", s=100, edgecolors="black", zorder=3)

for _, fila in df.iterrows():
    ax.annotate(fila["nombre"], (fila["horas_estudio"], fila["nota"]),
                textcoords="offset points", xytext=(5, 5), fontsize=9)

ax.set_title("Relación entre horas de estudio y nota final", fontsize=14, fontweight="bold")
ax.set_xlabel("Horas de estudio")
ax.set_ylabel("Nota final")
ax.set_ylim(0, 10)
ax.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

```
